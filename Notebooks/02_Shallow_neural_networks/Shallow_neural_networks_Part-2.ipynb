# Notebook 2 -- Shallow neural networks

The purpose of this notebook is to gain some familiarity with shallow neural networks with 1D inputs. It works through an example similar to figure and experiments with different activation functions.

```python
# Imports math library
import numpy as np
# Imports plotting library
import matplotlib.pyplot as plt
```

Let's first construct the shallow neural network with one input, three hidden units, and one output.

## Network architecture: 1 input -> 3 hidden units (ReLU) -> 1 output



**Layer 1 (hidden, 3 units):**

$$\text{pre}_k = b1_k + w1_k \cdot x \qquad \text{act}_k = \text{ReLU}(\text{pre}_k), \quad k=1,2,3$$

**Layer 2 (output, 1 unit):**

$$y = b2 + w2_1\cdot\text{act}_1 + w2_2\cdot\text{act}_2 + w2_3\cdot\text{act}_3$$


```python
# Define the Rectified Linear Unit (ReLU) function
def ReLU(preactivation):
  # Clip everything below 0 to 0; leave everything above 0 unchanged
  activation = preactivation.clip(min=0)
  return activation
```

```python
# Make an array of inputs
z = np.arange(-5,5,0.1)
RelU_z = ReLU(z)

# Plot the ReLU function
fig, ax = plt.subplots()
ax.plot(z,RelU_z,'r-')
ax.set_xlim([-5,5]);ax.set_ylim([-5,5])
ax.set_xlabel('z'); ax.set_ylabel('ReLU[z]')
ax.set_aspect('equal')
plt.show()
```

```python
# Define a shallow neural network with, one input, one output, and three hidden units
def shallow_1_1_3(x, activation_fn, b2,w2_1,w2_2,w2_3, b1_1, w1_1, b1_2, w1_2, b1_3, w1_3):
  # Compute the three initial lines from the weight (w) and bias (b) parameters
  # (equations at bottom of figure a-c). These are the preactivations
  pre_1 = b1_1 + w1_1 * x
  pre_2 = b1_2 + w1_2 * x
  pre_3 = b1_3 + w1_3 * x

  # Pass these through the ReLU function to compute the activations as in
  # figure d-f
  act_1 = activation_fn(pre_1)
  act_2 = activation_fn(pre_2)
  act_3 = activation_fn(pre_3)

  # Weight the activations using w2_1, w2_2 and w2_3
  # To create the equivalent of figure g-i
  w_act_1 = w2_1 * act_1
  w_act_2 = w2_2 * act_2
  w_act_3 = w2_3 * act_3

  # Combine the weighted activations and add
  # b2 to create the output as in figure j
  y = b2 + w_act_1 + w_act_2 + w_act_3

  # Return everything we have calculated
  return y, pre_1, pre_2, pre_3, act_1, act_2, act_3, w_act_1, w_act_2, w_act_3
```

```python
# Plot the shallow neural network.  We'll assume input in is range [0,1] and output [-1,1]
# If the plot_all flag is set to true, then we'll plot all the intermediate stages as in Figure
def plot_neural(x, y, pre_1, pre_2, pre_3, act_1, act_2, act_3, w_act_1, w_act_2, w_act_3, plot_all=False, x_data=None, y_data=None):

  # Plot intermediate plots if flag set
  if plot_all:
    fig, ax = plt.subplots(3,3)
    fig.set_size_inches(8.5, 8.5)
    fig.tight_layout(pad=3.0)
    ax[0,0].plot(x,pre_1,'r-'); ax[0,0].set_ylabel('Preactivation')
    ax[0,1].plot(x,pre_2,'b-'); ax[0,1].set_ylabel('Preactivation')
    ax[0,2].plot(x,pre_3,'g-'); ax[0,2].set_ylabel('Preactivation')
    ax[1,0].plot(x,act_1,'r-'); ax[1,0].set_ylabel('Activation')
    ax[1,1].plot(x,act_2,'b-'); ax[1,1].set_ylabel('Activation')
    ax[1,2].plot(x,act_3,'g-'); ax[1,2].set_ylabel('Activation')
    ax[2,0].plot(x,w_act_1,'r-'); ax[2,0].set_ylabel('Weighted Act')
    ax[2,1].plot(x,w_act_2,'b-'); ax[2,1].set_ylabel('Weighted Act')
    ax[2,2].plot(x,w_act_3,'g-'); ax[2,2].set_ylabel('Weighted Act')

    for plot_y in range(3):
      for plot_x in range(3):
        ax[plot_y,plot_x].set_xlim([0,1]);ax[plot_x,plot_y].set_ylim([-1,1])
        ax[plot_y,plot_x].set_aspect(0.5)
      ax[2,plot_y].set_xlabel('Input, $x$');
    plt.show()

  fig, ax = plt.subplots()
  ax.plot(x,y)
  ax.set_xlabel('Input, $x$'); ax.set_ylabel('Output, $y$')
  ax.set_xlim([0,1]);ax.set_ylim([-1,1])
  ax.set_aspect(0.5)
  if x_data is not None:
    ax.plot(x_data, y_data, 'mo')
    for i in range(len(x_data)):
      ax.plot(x_data[i], y_data[i],)
  plt.show()
```

```python
# Now lets define some parameters and run the neural network
b1_1 =  0.3 ; w1_1 = -1.0
b1_2 = -1.0  ; w1_2 = 2.0
b1_3 = -0.5  ; w1_3 = 0.65
b2 = -0.3; w2_1 = 2.0; w2_2 = -1.0; w2_3 = 7.0

# Define a range of input values
x = np.arange(0,1,0.01)

# We run the neural network for each of these input values
y, pre_1, pre_2, pre_3, act_1, act_2, act_3, w_act_1, w_act_2, w_act_3 = \
    shallow_1_1_3(x, ReLU, b2,w2_1,w2_2,w2_3, b1_1, w1_1, b1_2, w1_2, b1_3, w1_3)
# And then plot it
plot_neural(x, y, pre_1, pre_2, pre_3, act_1, act_2, act_3, w_act_1, w_act_2, w_act_3, plot_all=True)
```

# Least squares loss

Now let's consider fitting the network to data.  First we need to define the loss function.  We'll use the least squares loss:

\begin{equation}
L[\mathbf{w},\mathbf{b}] = \sum_{i=1}^{I}(y_{i}-\text{f}[x_{i},\mathbf{w},\mathbf{b}])^2
\end{equation}

where $(x_i,y_i)$ is an input/output training pair and $\text{f}[\bullet,\mathbf{w},\mathbf{b}]$ is the neural network with parameters $(\mathbf{w},\mathbf{b})$.  The first term in the brackets is the ground truth output and the second term is the prediction of the model

```python
# Least squares function
def least_squares_loss(y_train, y_predict):
  # Sum of squared differences between the real values of y and the
  # predicted values from the model f[x_i,w,b]

  loss = np.sum((y_train - y_predict) ** 2)

  return loss
```
