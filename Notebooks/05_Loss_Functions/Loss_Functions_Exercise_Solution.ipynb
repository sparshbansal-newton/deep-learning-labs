**Task 1: Delivery Time Predictions (MSE)**

AI Town's delivery bot predicted how many **minutes** 5 orders would take. Compare the predictions to what actually happened using **Mean Squared Error** — it punishes big misses hard.

```python
predicted = torch.tensor([30., 25., 40., 20., 35.])   # model's guess (minutes)
actual    = torch.tensor([28., 30., 38., 26., 35.])   # real time (minutes)

#1 Compute MSE by hand  (mean of (actual - predicted) squared)
mse_manual = torch.mean((actual - predicted) ** 2)

#2 Compute the same thing with PyTorch's built-in loss
mse_builtin = nn.MSELoss()(predicted, actual)

print("Manual MSE :", mse_manual)     # tensor(13.8000)
print("Built-in MSE:", mse_builtin)   # tensor(13.8000)  -> they match ✅
```

**Task 2: Surge Pricing & a Rogue Fare (MAE vs MSE)**

A ride app in Mumbai estimated 5 fares (in ₹). One trip went through a toll and became a huge **outlier**. Compute **MAE** (treats every error equally) and compare it to MSE.

```python
fare_pred = torch.tensor([120., 150., 200., 90., 300.])
fare_true = torch.tensor([130., 145., 210., 95., 900.])   # last one is the outlier 🧨

#1 Compute MAE by hand  (mean of the ABSOLUTE errors)
mae_manual = torch.mean(torch.abs(fare_true - fare_pred))

#2 Compute MAE and MSE with the built-in losses
mae_builtin = nn.L1Loss()(fare_pred, fare_true)
mse_builtin = nn.MSELoss()(fare_pred, fare_true)

print("MAE:", mae_builtin, " | MSE:", mse_builtin)
# MAE: tensor(126.)  |  MSE: tensor(72050.)  -> the squared error explodes on the outlier

#3 Which loss got dragged up more by the outlier?
more_affected = "MSE"
print("More affected by the outlier:", more_affected)
```

**Task 3: Tumour Screening (Binary Cross-Entropy)**

AI Town Hospital's model outputs a **probability** that a scan is benign (1) or malignant (0). Score it with **BCE** — remember, `nn.BCELoss` expects *probabilities*, not raw scores.

```python
probs  = torch.tensor([0.9, 0.2, 0.8, 0.3, 0.6])   # model's probability of "benign"
labels = torch.tensor([1., 0., 1., 0., 1.])        # ground truth (1=benign, 0=malignant)

#1 Compute the Binary Cross-Entropy loss
bce_loss = nn.BCELoss()(probs, labels)

print("BCE loss:", bce_loss)   # tensor(0.2838)
```

**Task 4: The Logit Trap (BCEWithLogitsLoss)**

A newer model outputs **logits** (raw scores, any real number) instead of probabilities. `nn.BCEWithLogitsLoss` applies the sigmoid *for you* — so feed it the **logits**, never the probabilities.

```python
logits = torch.tensor([2.0, -1.5, 0.5, -0.2, 1.0])   # raw model outputs
labels = torch.tensor([1.,  0.,   1.,  0.,   1.])

#1 Turn the logits into probabilities yourself
probs_from_logits = torch.sigmoid(logits)

#2 Compute the loss the CORRECT way: pass the LOGITS to BCEWithLogitsLoss
loss_correct = nn.BCEWithLogitsLoss()(logits, labels)

print("Probabilities:", probs_from_logits)   # tensor([0.8808, 0.1824, 0.6225, 0.4502, 0.7311])
print("Loss (from logits):", loss_correct)   # tensor(0.3428)

# ⚠️ Common mistake: nn.BCEWithLogitsLoss()(probs_from_logits, labels) would sigmoid AGAIN and be wrong.
```

**Task 5: Handwritten Digit Sorter (Cross-Entropy)**

AI Town's post office sorts pincodes into **3 bins**. For 4 images the model gives a row of 3 raw scores (logits). Use **Cross-Entropy** — it wants **raw logits** and **integer class labels** (it does softmax internally, so don't softmax first!).

```python
logits = torch.tensor([[2.0, 0.5, 0.1],
                       [0.3, 1.8, 0.2],
                       [0.4, 0.4, 2.5],
                       [1.5, 0.2, 0.3]])          # shape [4, 3]
targets = torch.tensor([0, 1, 2, 0])              # correct class per image (int64)

#1 Compute the Cross-Entropy loss  (raw logits + integer targets)
ce_loss = nn.CrossEntropyLoss()(logits, targets)

#2 Which class does the model currently predict for each image?
predictions = torch.argmax(logits, dim=1)

print("Cross-Entropy loss:", ce_loss)      # tensor(0.3359)
print("Predicted classes :", predictions)  # tensor([0, 1, 2, 0])
print("True classes      :", targets)      # tensor([0, 1, 2, 0])  -> all correct here ✅
```
