**Task5: The Reset Button (Gradient Accumulation)**

PyTorch **adds** each new gradient onto whatever is already in `.grad`. Handy for mini-batches, disastrous in a training loop if you forget to clear it. Here you'll watch a gradient accumulate, then reset it.

```python
#1 Create x = 2.0 (tracked). Compute y = x**2, backprop, and print x.grad

x = torch.tensor(__, requires_grad=__)
y = x ** 2
y.backward()
print(x.grad)     # first gradient
```

```python
#2 WITHOUT clearing, recompute y = x**2, backprop again, and print x.grad
# Hint: it should be DOUBLE the previous value — gradients accumulate

y = x ** 2
y.__()
print(x.grad)     # accumulated gradient
```

```python
#3 Reset the accumulated gradient back to zero, then print x.grad
# Hint: use the in-place x.grad.zero_()

x.grad.__()
print(x.grad)
```
